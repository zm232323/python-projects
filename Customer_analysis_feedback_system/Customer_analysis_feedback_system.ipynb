Customer Feedback analysis system


```python
#take number of inputs
input_feedbacks = int(input("Enter number of feedback: "))

positive_feedback = []
negative_feedback = []
neutral_feedback = []

#Take feedbacks and check positive,negative,neutral feedbacks and store in list
for i in range(input_feedbacks):
  feedbacks = input("Enter feedback: ").lower().strip()
  if("good" in feedbacks or "excellent" in feedbacks or "amazing" in feedbacks or "fast" in feedbacks or "great" in feedbacks or "like" in feedbacks):
    positive_feedback.append(feedbacks)
  elif("bad" in feedbacks or "poor" in feedbacks or "rude" in feedbacks or "slow" in feedbacks or "worst" in feedbacks):
    negative_feedback.append(feedbacks)
  else:
    neutral_feedback.append(feedbacks)

#Display function is used to display the result
def display():
  print("\n---- Feedback analysis result ----")
  print(f"Total feedbacks: {input_feedbacks}")
  print(f"Positive feedback: {len(positive_feedback)}")
  for j in range(len(positive_feedback)):
    print(f"{j+1}. {positive_feedback[j]}")
  print(f"Negative feedback: {len(negative_feedback)}")
  for j in range(len(negative_feedback)):
    print(f"{j+1}. {negative_feedback[j]}")
  print(f"Neutral feedback: {len(neutral_feedback)}")
  for j in range(len(neutral_feedback)):
    print(f"{j+1}. {neutral_feedback[j]}")

  print(f"\n____insights_____")
  print(f"Positive count: {len(positive_feedback)}")
  print(f"Negative count: {len(negative_feedback)}")
  print(f"Neutral count: {len(neutral_feedback)}")

#Finally call the display functiond
display()

#Search by keywords
input_keyword = input("Enter a keyword to search: ").lower()

print("\nResults:")

found = False

all_feedbacks = positive_feedback + negative_feedback + neutral_feedback

for fb in all_feedbacks:
    if input_keyword in fb:
        print(f"- {fb}")
        found = True

if not found:
    print("No keyword match!")
```
