```python
diseases = {
    "Flu": ["fever", "cough", "fatigue"],
    "Common Cold": ["cough", "sneezing", "headache"],
    "Malaria": ["fever", "chills", "vomiting"]
}

while True:

    user_symptoms = []

    print("\nEnter 3 Symptoms\n")

    for i in range(3):
        symptoms = input("Enter the symptoms: ").lower()
        user_symptoms.append(symptoms)

    flu_match_count = 0
    common_cold_match_count = 0
    malaria_match_count = 0

    for symptom in user_symptoms:

        if symptom in diseases["Flu"]:
            flu_match_count += 1

        if symptom in diseases["Common Cold"]:
            common_cold_match_count += 1

        if symptom in diseases["Malaria"]:
            malaria_match_count += 1

    print("\n--- Disease Match Result ---\n")

    # Flu Result
    if flu_match_count == 3:
        print("Flu -> High Match")
    elif flu_match_count == 2:
        print("Flu -> Medium Match")
    elif flu_match_count == 1:
        print("Flu -> Low Match")
    else:
        print("Flu -> No Match")

    # Common Cold Result
    if common_cold_match_count == 3:
        print("Common Cold -> High Match")
    elif common_cold_match_count == 2:
        print("Common Cold -> Medium Match")
    elif common_cold_match_count == 1:
        print("Common Cold -> Low Match")
    else:
        print("Common Cold -> No Match")

    # Malaria Result
    if malaria_match_count == 3:
        print("Malaria -> High Match")
    elif malaria_match_count == 2:
        print("Malaria -> Medium Match")
    elif malaria_match_count == 1:
        print("Malaria -> Low Match")
    else:
        print("Malaria -> No Match")

    print("\n--- Most Likely Disease ---\n")

    # Most Likely Disease
    if flu_match_count > common_cold_match_count and flu_match_count > malaria_match_count:
        print("Most Likely Disease -> Flu")

    elif common_cold_match_count > flu_match_count and common_cold_match_count > malaria_match_count:
        print("Most Likely Disease -> Common Cold")

    elif malaria_match_count > flu_match_count and malaria_match_count > common_cold_match_count:
        print("Most Likely Disease -> Malaria")

    elif flu_match_count == common_cold_match_count and flu_match_count > malaria_match_count:
        print("Most Likely Diseases -> Flu and Common Cold")

    elif flu_match_count == malaria_match_count and flu_match_count > common_cold_match_count:
        print("Most Likely Diseases -> Flu and Malaria")

    elif common_cold_match_count == malaria_match_count and common_cold_match_count > flu_match_count:
        print("Most Likely Diseases -> Common Cold and Malaria")

    else:
        print("Unable to predict disease clearly")

    again = input("\nDo you want to check again? (yes/no): ").lower()

    if again == "no":
        print("Thank You")
        break
```
