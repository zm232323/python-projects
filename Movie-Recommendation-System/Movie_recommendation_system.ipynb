```python
print("Welcome To Movie Recommendation System")
print("--------------------------------------")

movies = [
    ("Gravity", "Sci-Fi", 8.1, 13),
    ("Avatar", "Sci-Fi", 8.0, 13),
    ("Mad Max", "Action", 8.3, 16),
    ("Extraction", "Action", 8.1, 18),
    ("Fast and Furious", "Action", 8.0, 13),
    ("Me Before You", "Romance", 7.9, 13),
    ("Love Again", "Romance", 7.8, 13),
    ("Coco", "Animation", 8.4, 7),
    ("Cars", "Animation", 8.0, 5),
    ("Minions", "Animation", 7.9, 5)
]

user_name = input("\nEnter Your Name: ")
user_age = int(input("Enter Your Age: "))
preferred_genre = input("Enter Preferred Genre(Animation,Action,Sci-Fi,Romance): ")

print("\nUser Information")
print("Name:", user_name)
print("Age:", user_age)
print("Genre:", preferred_genre)

print("\nRecommended Movies For", user_name)
print("-----------------------------------")

movie_count = 0

for movie in movies:

    movie_name = movie[0]
    movie_genre = movie[1]
    movie_rating = movie[2]
    movie_age = movie[3]

    if movie_genre.lower() == preferred_genre.lower():

        if user_age >= movie_age:

            if movie_rating >= 8:

                print("Movie:", movie_name)
                print("Rating:", movie_rating)
                print()

                movie_count = movie_count + 1

if movie_count == 0:
    print("No suitable movie found")

print("Total Recommended Movies:", movie_count)

print("\nThank You For Using The System")
```
